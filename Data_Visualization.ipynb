```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
plt.plot(x, y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Line Chart Example")
plt.show()
```

```python
import matplotlib.pyplot as plt
data = [10, 12, 15, 18, 20, 22, 25, 30, 35, 50]
plt.boxplot(data)
plt.title("Box Plot Example")
plt.xlabel("Data")
plt.ylabel("Values")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
data = [10, 12, 15, 18, 20, 22, 25, 30, 35, 50]
sns.boxplot(x=data)
plt.title("Box Plot")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
subjects = ["DMPM", "GAA", "ESE", "AML", "OB"]
students = [40, 30, 50, 50 , 25]

plt.bar(subjects, students)

plt.xlabel("Subjects")
plt.ylabel("Number of Students")
plt.title("Bar Chart Example")

plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ["Python", "Java", "C++", "JavaScript", "Html"]
values = [40, 25, 20, 19 ,10]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%"
)

plt.title("Programming Language Usage")
plt.show()
```

```python
data = ["A", "B", "A", "C", "B", "A", "C", "B"]

sns.countplot(x=data)
plt.show()
```

```python
height = [150, 160, 165, 170, 175, 180]
weight = [50, 55, 60, 65, 70, 80]

plt.scatter(height, weight)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```

```python
data = [10, 12, 15, 15, 18, 20, 20, 22, 25, 30]
plt.hist(data, bins=5)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```
