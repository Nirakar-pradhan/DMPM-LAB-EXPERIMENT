```python
'''MIN-MAX'''
import numpy as np
a = np.array([10, 20, 30, 40, 50])
min_value = np.min(a)
max_value = np.max(a)
min_max = (a - min_value) / (max_value - min_value)
print("Original:", a)
print("Min-Max:", min_max)
```

```python
'''Z-SCORE'''
import numpy as np
a = np.array([10, 20, 30, 40, 50])
mean = np.mean(a)
std = np.std(a)
z_score = (a - mean) / std
print("Original:", a)
print("Mean:", mean)
print("Standard Deviation:", std)
print("Z-Score:", z_score)
```

```python
'''DECIMAL SCALLING'''
import numpy as np
a = np.array([10, 20, 30, 40, 50])
max_abs = np.max(np.abs(a))
j = len(str(int(max_abs)))
decimal_scaling = a / (10 ** j)
print("Original:", a)
print("Decimal Scaling:", decimal_scaling)
```

```python
'''CHI-SQUARE TEST'''
import numpy as np
from scipy.stats import chi2_contingency
# Observed data
a = np.array([
    [30, 20],
    [10, 40]
])
chi2, p, dof, expected = chi2_contingency(a)
print("Chi-Square:", chi2)
print("P-value:", p)
print("Degrees of Freedom:", dof)
print("\nExpected Frequencies:")
print(expected)
if p < 0.05:
    print("Reject H0: Variables are related.")
else:
    print("Fail to reject H0: Variables are independent.")
```
