```python
import math

def mean(a):
    return sum(a) / len(a)

def median(a):
    a.sort()
    n = len(a)
    return (a[n//2] + a[(n-1)//2]) / 2

def mode(a):
    return max(a, key=a.count)

def variance(a):
    m = mean(a)
    return sum((i - m) ** 2 for i in a) / len(a)

def std_dev(a):
    return math.sqrt(variance(a))

a = list(map(int, input("Enter numbers: ").split()))

print("Mean =", mean(a))
print("Median =", median(a))
print("Mode =", mode(a))
print("Variance =", variance(a))
print("Standard Deviation =", std_dev(a))
```

```python
def mean_calc(list_1):
  sum=0
  count=0
  for i in list_1:
    sum+=i
    count+=1
  return(sum/count)
mean_sol=mean_calc([1,2,3,4,5])
print(mean_sol)

```

```python
def median_calc(list_2):
  count=0
  for i in list_2:
    count+=1
  list_2.sort()
  if(count%2==0):
    return((list_2[count//2]+list_2[(count//2)-1])/2)
  else:
    return(list_2[count//2])
median_sol=median_calc([1,2,3,4,5,6])
print(median_sol)
```

```python
def mode_calc(list_3,n):
  if (n in list_3):
    count=0
    for numbers in list_3:
      if(numbers==n):
        count+=1
    return(count)
  else:
    return(0)
mode_sol=mode_calc([1,2,3,4,5,6,2,7,8,4],2)
print(mode_sol)
```

```python
def variance_calc(list_4):
  sum=0
  for i in list_4:
    sum+=i
  mean=sum/len(list_4)
  sum=0
  for i in list_4:
    sum+=(i-mean)**2
  return(sum/len(list_4))
variance_sol=variance_calc([1,2,3,4,5,6,2,7,8,4])
print(variance_sol)
```

```python
def standard_dev(list_5):
  sum=0
  for i in list_5:
    sum+=i
  mean=sum/len(list_5)
  sum=0
  for i in list_5:
    sum+=(i-mean)**2
  return(sum/len(list_5))**0.5
standard_dev_sol=standard_dev([1,2,3,4,5,6,2,7,8,4])
print(standard_dev_sol)
```

```python
def varience_calc(list_4):
  var=standard_dev(list_4)
  return var**2
var=varience_calc([1,2,3,4,5,12,5,7,8])
print(var)
```
